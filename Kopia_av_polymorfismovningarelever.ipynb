### Övning 1 — Repetition: override ger polymorfism

Skapa en klass `Instrument` med attributet `namn` och en metod `spela()` som skriver ut `"<namn> spelas."`

Skapa `Gitarr(Instrument)` som overridar `spela()` till `"<namn> knäpper strängarna."`, och `Trumma(Instrument)` som overridar den till `"<namn> trummar."`

Loopa igenom en lista med minst en av varje (inklusive ett vanligt `Instrument`-objekt) och anropa `spela()` på alla.

```python
class Instrument:
    def __init__(self, namn):
        self.namn = namn

    def spela(self):
        print(f"{self.namn} spelas.")


class Gitarr(Instrument):
    def spela(self):
        print(f"{self.namn} knäpper strängarna.")


class Trumma(Instrument):
    def spela(self):
        print(f"{self.namn} trummar.")


instrument_lista = [Instrument("Bas"), Gitarr("Akustisk gitarr"), Trumma("Virveltrumma")]

for instrument in instrument_lista:
    instrument.spela()
```

### Övning 4 — Polymorfism via en gemensam funktion

Skapa `Instrument`, `Gitarr` och `Trumma` (som i övning 1). Skriv en funktion `spela_instrument(instrument)` som bara anropar `instrument.spela()`.

Anropa funktionen med ett `Gitarr`-objekt och ett `Trumma`-objekt.

```python
class Instrument:
    def __init__(self, namn):
        self.namn = namn

    def spela(self):
        print(f"{self.namn} spelas.")


class Gitarr(Instrument):
    def spela(self):
        print(f"{self.namn} knäpper strängarna.")


class Trumma(Instrument):
    def spela(self):
        print(f"{self.namn} trummar.")


def spela_instrument(instrument):
    instrument.spela()


spela_instrument(Gitarr("Akustisk gitarr"))
spela_instrument(Trumma("Virveltrumma"))
```

### Övning 5 — Duck typing

Skapa två klasser, `Telefon` och `Dörrklocka`, som INTE ärver från någon gemensam superklass. Ge båda en metod `ring()` som skriver ut ett lämpligt meddelande (t.ex. `"Telefonen ringer!"` respektive `"Dörrklockan ringer!"`).

Loopa igenom en lista med ett objekt av vardera klass och anropa `ring()` på båda.

```python
class Telefon:
    def ring(self):
        print("Telefonen ringer!")


class Dörrklocka:
    def ring(self):
        print("Dörrklockan ringer!")


saker_som_ringer = [Telefon(), Dörrklocka()]

for sak in saker_som_ringer:
    sak.ring()
```

### Övning 6 — Generisk funktion + duck typing

Skapa `Instrument` med `spela()`, och funktionen `spela_instrument(instrument)` (som i övning 4).

Skapa sedan en helt ny klass `Radio` som INTE ärver från `Instrument`, men som också har en `spela()`-metod (t.ex. `"Radion spelar musik."`).

Anropa `spela_instrument()` med ett `Radio`-objekt och visa att det fungerar ändå.

### Övning 7 — Överlagra `__add__`

Skapa en klass `Poang` med attributet `varde`. Lägg till en `__add__`-metod så att två `Poang`-objekt kan adderas ihop med `+` (resultatet ska vara ett nytt `Poang`-objekt med summan av värdena).

Skapa två `Poang`-objekt, addera dem, och skriv ut resultatets `varde`.

```python
class Poang:
    def __init__(self, varde):
        self.varde = varde

    def __add__(self, other):
        return Poang(self.varde + other.varde)


poang1 = Poang(12)
poang2 = Poang(8)
total = poang1 + poang2

print(total.varde)
```

### Övning 8 — Överlagra `__len__`

Skapa en klass `Spellista` med attributet `latar` (en lista med låtnamn). Lägg till en `__len__`-metod som gör att `len(min_spellista)` returnerar antalet låtar.

Skapa en `Spellista` med minst 3 låtar och skriv ut `len()` av den.

```python
class Spellista:
    def __init__(self, latar):
        self.latar = latar

    def __len__(self):
        return len(self.latar)


min_spellista = Spellista(["Låt ett", "Låt två", "Låt tre"])
print(len(min_spellista))
```

### Övning 9 — `__str__` är också polymorfism

Skapa två klasser, `Biljett` och `Kvitto`, båda med en egen `__str__`-metod som returnerar en läsbar textrad (fritt vad de innehåller, t.ex. biljettens resmål respektive kvittots totalsumma).

Skriv ut ett objekt av vardera klass i en loop med `print()`.

```python
class Biljett:
    def __init__(self, resmal):
        self.resmal = resmal

    def __str__(self):
        return f"Biljett till {self.resmal}"


class Kvitto:
    def __init__(self, totalsumma):
        self.totalsumma = totalsumma

    def __str__(self):
        return f"Kvitto: totalsumman är {self.totalsumma} kr"


utskrifter = [Biljett("Stockholm"), Kvitto(249)]

for objekt in utskrifter:
    print(objekt)
```

### Övning 10 — Sammanfattande övning: blandat band

Skapa `Instrument`, `Gitarr` och `Trumma` (som i övning 1). Skapa också en duck-typad klass `Radio` (som i övning 6) med en `spela()`-metod, men som INTE ärver från `Instrument`.

Skapa en lista `bandet` med minst en `Gitarr`, en `Trumma` och en `Radio`. Loopa igenom listan, anropa `spela()` på varje, och använd `isinstance()` för att skriva ut om objektet är ett "riktigt" instrument eller inte.

```python
class Instrument:
    def __init__(self, namn):
        self.namn = namn

    def spela(self):
        print(f"{self.namn} spelas.")


class Gitarr(Instrument):
    def spela(self):
        print(f"{self.namn} knäpper strängarna.")


class Trumma(Instrument):
    def spela(self):
        print(f"{self.namn} trummar.")


class Radio:
    def spela(self):
        print("Radion spelar musik.")


bandet = [Gitarr("Elgitarr"), Trumma("Bastrumma"), Radio()]

for instrument in bandet:
    instrument.spela()
    if isinstance(instrument, Instrument):
        print("Det här är ett riktigt instrument.")
    else:
        print("Det här är inte ett riktigt instrument.")
```
